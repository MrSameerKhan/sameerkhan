### Bag of Words

```python
from sklearn.feature_extraction.text import CountVectorizer

corpus = [
    "He is a good boy",
    "She is a good girl",
    "Boy and girl are good"
]

vectorizer = CountVectorizer(stop_words='english')
X = vectorizer.fit_transform(corpus)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```
