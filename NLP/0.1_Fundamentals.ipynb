### Text Preprocessing

1. Text Cleaning

Definition: Removing unwanted characters, symbols, or formatting to normalize raw text.
Why: Raw text (tweets, reviews, medical notes) is noisy → cleaning improves model accuracy.

Steps:
Lowercasing
Removing punctuation, special symbols
Handling extra spaces, HTML tags, numbers, emojis (depending on task)

Real-world use case:
In sentiment analysis of tweets, removing hashtags, mentions, links, and emojis before feeding into a model.
In healthcare clinical notes, removing patient IDs and symbols.

```python
import re

text = "Hello!!! This is NLP, 2025 🚀. <br> Let's start." #re.sub search and replace
cleaned = re.sub(r"<.*?>", "", text)   # remove HTML
cleaned = re.sub(r"[^a-zA-Z\s]", "", cleaned)  # remove punctuation/numbers/emojis
cleaned = cleaned.lower().strip()

print(cleaned)
# Output: "hello this is nlp lets start"
```

2. Tokenization

Definition: Splitting text into smaller units (tokens) like words, subwords, or characters.
Why: Models cannot directly understand text → tokens act as the input units.

Types:
Word-level: Split on spaces → "I love AI" → ["I", "love", "AI"]
Problem: Out-of-vocabulary (OOV) words.
Subword-level (BPE, WordPiece, SentencePiece)
"unhappiness" → ["un", "happiness"]
Reduces OOV problem → used in BERT, GPT.
Character-level: "AI" → ["A", "I"]
Used in spelling correction, OCR.

Real-world use case:
Search engines → split queries into tokens for indexing.
BERT/GPT → subword tokenization to efficiently handle huge vocabularies.

```python
import os
import warnings
import nltk
from transformers import AutoTokenizer

# Suppress warnings & progress bars
warnings.filterwarnings("ignore")
os.environ["HF_HUB_DISABLE_PROGRESS_BARS"] = "1"

# Download quietly
nltk.download('punkt', quiet=True) #C:\Users\<your-username>\AppData\Roaming\nltk_data\punkt\english.pickle
nltk.download('punkt_tab', quiet=True)

text = "I love NLP and Transformers!"

# Word tokenization
print(nltk.word_tokenize(text))  

# Subword tokenization (BERT WordPiece)
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased") #C:\Users\samee\.cache\huggingface\hub\models--bert-base-uncased\snapshots\
print(tokenizer.tokenize(text))
```

3. Stop Words Removal

Definition: Removing frequently occurring but less informative words like “the”, “is”, “and”.
Why: They may add noise in classical ML models (e.g., TF-IDF, BoW).
Real-world use case:
In search ranking, ignoring stop words prevents unnecessary matches.
In clinical text mining, removing “patient was given” type fillers.
👉 Note: In transformer models, stop words are usually kept, since context matters.

```python
from nltk.corpus import stopwords
nltk.download('stopwords', quiet=True) #nltk_data\corpora\stopwords

text = ["this", "is", "a", "sample", "sentence"]
filtered = [w for w in text if w not in stopwords.words("english")]
print(filtered)
# ['sample', 'sentence']
```

4. Stemming vs Lemmatization
Definition: Reducing words to root/base forms.
Stemming: Rule-based chopping → "playing" → "play" (sometimes "studies" → "studi")
Lemmatization: Dictionary + grammar-based → "studies" → "study"

Real-world use case:
Search engines use lemmatization (better results).
Topic modeling may use stemming (faster, but less precise).
👉 Interview tip: Say lemmatization is preferred in modern NLP pipelines for accuracy.

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer
nltk.download('wordnet', quiet = True)

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

print(stemmer.stem("studies"))   # studi
print(lemmatizer.lemmatize("studies"))  # study
```

7. Named Entity Recognition (NER)

Definition: Detecting named entities (persons, places, dates, organizations).
Why: Critical for extracting structured knowledge.

Real-world use case:

```python
import spacy
nlp = spacy.load("en_core_web_sm")

doc = nlp("Elon Musk founded SpaceX in 2002, headquartered in California.")
for ent in doc.ents:
    print(ent.text, ent.label_)
# Elon Musk PERSON
# SpaceX ORG
# 2002 DATE
# California GPE
```

1. Bag of Words (BoW)

Definition:
BoW represents text as a “bag” of words, ignoring grammar and order, just counting word frequency.
A feature extration method in NLP that converts text into a fixed vector based on the occurence of words, ignoring grammer and word order

How it works:
Build a vocabulary of all unique words in corpus.
Represent each document as a vector of word counts.

Example:
Corpus:
Doc1: "I love NLP"
Doc2: "I love AI"

Vocabulary = [I, love, NLP, AI]

Vectors:
Doc1 → [1, 1, 1, 0]
Doc2 → [1, 1, 0, 1]

Real-world use case:
Early spam detection (counting spammy words).
Document similarity in search engines (before embeddings).

Limitations:
Ignores order of words (semantics lost).
Produces high-dimensional sparse vectors.
Treats all words equally (no weighting).

```python
from sklearn.feature_extraction.text import CountVectorizer

docs = ["I love NLP", "I love AI"]
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(docs)

print(vectorizer.get_feature_names_out())
# ['ai' 'love' 'nlp']
print(X.toarray())
# [[0 1 1]   -> "I love NLP"
#  [1 1 0]]  -> "I love AI"
```

2. TF-IDF (Term Frequency – Inverse Document Frequency)

Definition:
TF-IDF improves BoW by down-weighting common words and up-weighting rare but important words.

TF(Term Frequency) -> how often a word appears in a document
IDF(Inverse Document Frequency) -> how rare the word is across all documents

Goal: Give higher weight to important & rare words and lower weight to common words

Formula:
TF (Term Frequency):TF(t,d)=count of term t in document d​/total terms in document d
IDF (Inverse Document Frequency):IDF(t) = log(N/DF(t))

where
N = total number of documents
DF(t) = = number of documents containing term t
Combined = TF-IDF(t,d) = TF(t,d) * IDF(t)

Intuition:
Very frequent words across all documents (“the”, “is”, “and”) → low IDF → low weight.
Rare, domain-specific words (“transformer”, “diabetes”, “Elon Musk”) → high IDF → high weight.

Example Output (weights vary):
Sentence corpus → "I love NLP", "AI is the future of NLP"
“NLP”, “AI”, “future” → high weight
“I”, “the”, “is” → low weight

Real-world use cases:
Search engines → Google’s early ranking systems (before embeddings).
News classification → detect category terms like “sports”, “politics”.
Legal/medical retrieval → highlight domain-specific terminology.
Spam filtering → spammy words (e.g., “free”, “win”, “lottery”) get weighted higher.

Limitations:
Still sparse & high-dimensional.
No semantic meaning → synonyms treated differently (doctor ≠ physician).
Doesn’t capture word order (context lost).

Extra Interview Points to Add:
Variants of IDF:
Smoothed IDF: log(1 + N/1+DF(t)) -> prevents divide by zero
Sublinear TF Scaling : use 1 + log(TF) instead of raw counts

Comparison with Embeddings:
TF-IDF is interpretable & fast, still used in production for baseline ranking.
Embeddings (Word2Vec, BERT) capture semantics, but heavier to compute.

When NOT to use TF-IDF:
In transformer-based NLP pipelines (contextual embeddings are superior).
When you need semantic similarity across synonyms.

So, for interviews, if they ask “Why TF-IDF over BoW?” → You answer: “It balances word frequency with importance, penalizes common words, and highlights rare informative terms.”

```python
from sklearn.feature_extraction.text import TfidfVectorizer

docs = ["I love NLP", "I love AI", "AI is the future of NLP"]
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(docs)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```

3. N-gram Models

Definition:
N-grams capture sequences of N words (instead of treating words independently).

Unigram (N=1): "I", "love", "AI"
Bigram (N=2): "I love", "love AI"
Trigram (N=3): "I love NLP"
4-gram (N=4): "This is the house"

Why:
Adds local context that BoW/TF-IDF miss.
Instead of treating each word in isolation, n-grams model how words co-occur.

Example:
Sentence: "I love NLP models"
Bigrams = ["I love", "love NLP", "NLP models"]
Trigrams = ["I love NLP", "love NLP models"]
Probability Estimation (Markov assumption):

P(w4|w1, w2, w3) = C(w1, w2, w3, w4)/C(w1, w2,w3)


E.g., from the rhyme:
P(house|this is the) = C(this is the house)/C(this is the) = 1/4


Real-world use cases:
Autocomplete → predicting next word using bigrams/trigrams.
Speech recognition → statistical language models.
Machine translation (early phrase-based MT systems).
Sentiment analysis → "not good" vs "good" (bigram helps).

Limitations:
Vocabulary explosion → large memory for high N.
Data sparsity → many n-grams never occur in training data (need smoothing like Laplace, Kneser-Ney).
No long-range dependencies → only looks at last (N-1) words.
Modern models (RNNs, Transformers) solve this.

Interview Angle:
Be ready to explain why we moved beyond N-grams.
Common Q: “How do you handle zero probabilities in N-gram models?” → Answer: Smoothing techniques (Laplace, Good-Turing, Kneser-Ney).
Another Q: “Why do transformers outperform N-grams?” → Because they capture global context and don’t suffer from sparsity.

```python
vectorizer = CountVectorizer(ngram_range=(2,2))  # bigrams
X = vectorizer.fit_transform(["I love NLP models", "NLP models are powerful"])
print(vectorizer.get_feature_names_out())
# ['love nlp' 'nlp models' 'models are' 'are powerful']
```

✅ Q1. Difference between BoW, TF-IDF, Word Embeddings?

BoW: counts words, sparse vectors.

TF-IDF: weighted counts, still sparse.

Embeddings: dense vectors, capture meaning.

✅ Q2. When to use BoW vs TF-IDF?

BoW: simple, when dataset small.

TF-IDF: better for classification/search tasks.

✅ Q3. Why are stop words sometimes not removed in TF-IDF?

Because IDF already penalizes common words like “the”, “is”.

✅ Q4. What’s a real drawback of n-grams?

Curse of dimensionality → vocabulary grows exponentially with n.

✅ Q5. How are these used in industry today?

Search engines, recommendation systems, spam filtering → TF-IDF still used.

Modern NLP prefers embeddings (Word2Vec, BERT), but TF-IDF is still useful for baseline models and keyword-based systems
