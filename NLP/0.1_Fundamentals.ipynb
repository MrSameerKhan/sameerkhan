### Text Preprocessing

1. Text Cleaning

Definition: Removing unwanted characters, symbols, or formatting to normalize raw text.
Why: Raw text (tweets, reviews, medical notes) is noisy → cleaning improves model accuracy.

Steps:

Lowercasing

Removing punctuation, special symbols

Handling extra spaces, HTML tags, numbers, emojis (depending on task)

```python
import re

text = "Hello!!! This is NLP, 2025 🚀. <br> Let's start." #re.sub search and replace
cleaned = re.sub(r"<.*?>", "", text)   # remove HTML
cleaned = re.sub(r"[^a-zA-Z\s]", "", cleaned)  # remove punctuation/numbers/emojis
cleaned = cleaned.lower().strip()

print(cleaned)
# Output: "hello this is nlp lets start"
```

2. Tokenization

Definition: Splitting text into smaller units (tokens) like words, subwords, or characters.
Why: Models cannot directly understand text → tokens act as the input units.

Types:

Word-level: Split on spaces → "I love AI" → ["I", "love", "AI"]

Problem: Out-of-vocabulary (OOV) words.

Subword-level (BPE, WordPiece, SentencePiece)

"unhappiness" → ["un", "happiness"]

Reduces OOV problem → used in BERT, GPT.

Character-level: "AI" → ["A", "I"]

Used in spelling correction, OCR.

```python
import os
import warnings
import nltk
from transformers import AutoTokenizer

# Suppress warnings & progress bars
warnings.filterwarnings("ignore")
os.environ["HF_HUB_DISABLE_PROGRESS_BARS"] = "1"

# Download quietly
nltk.download('punkt', quiet=True) #C:\Users\<your-username>\AppData\Roaming\nltk_data\punkt\english.pickle
nltk.download('punkt_tab', quiet=True)

text = "I love NLP and Transformers!"

# Word tokenization
print(nltk.word_tokenize(text))  

# Subword tokenization (BERT WordPiece)
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased") #C:\Users\samee\.cache\huggingface\hub\models--bert-base-uncased\snapshots\
print(tokenizer.tokenize(text))
```

3. Stop Words Removal

Definition: Removing frequently occurring but less informative words like “the”, “is”, “and”.
Why: They may add noise in classical ML models (e.g., TF-IDF, BoW).

```python
from nltk.corpus import stopwords
nltk.download('stopwords', quiet=True) #nltk_data\corpora\stopwords

text = ["this", "is", "a", "sample", "sentence"]
filtered = [w for w in text if w not in stopwords.words("english")]
print(filtered)
# ['sample', 'sentence']
```
