#                                                Word Embeddings


 Definition
Word embeddings are dense, low-dimensional vectors that represent words such that:
Words with similar meaning are close in vector space.
They capture semantic and syntactic relationships.

Unlike BoW/TF-IDF (sparse, thousands of dimensions), embeddings are:
Compact (50–300 dimensions typical)
Semantic (capture meaning, e.g., "king - man + woman ≈ queen")

# Word2Vec (Google, 2013)

Key Idea:
Train a shallow neural network on large text corpora.
Objective: words in similar contexts have similar embeddings.

 Training Architectures
CBOW (Continuous Bag of Words)
Predict target word from its context.
Example: "I ___ NLP" → predict "love".

Skip-gram
Predict context words from a given word.
Example: "love" → predict "I", "NLP".

Works better for rare words.
 Properties
Captures semantic similarity.
Famous analogy property:
"king - man + woman ≈ queen"
"Paris - France + Italy ≈ Rome"

```python
from gensim.models import Word2Vec

sentences = [["I", "love", "NLP"], ["I", "love", "AI"], ["AI", "is", "powerful"]]
model = Word2Vec(sentences, vector_size=50, window=3, min_count=1, sg=0)  # CBOW
print(model.wv["love"])  # embedding vector
print(model.wv.most_similar("AI"))
```

# GloVe (Stanford, 2014)

Key Idea:

Global Vectors: learns embeddings using matrix factorization of the word co-occurrence matrix.
Uses global corpus statistics rather than just local context windows (Word2Vec).

Intuition:
If "doctor" often co-occurs with "hospital", their vectors should be close.

Properties:
Pre-trained embeddings available (e.g., glove.6B.300d.txt).
Works very well on analogy tasks.

```python
from gensim.models import KeyedVectors

#Load pre-trained GloVe vectors (example: glove.6B.50d.txt converted to word2vec format)
model = KeyedVectors.load_word2vec_format("/Users/sameerkhan/Desktop/sameerkhan/data/nlp/glove.6B.50d.txt", binary=False, no_header=True)
print(model["hospital"])
```

# FastText (Facebook, 2016)

Key Idea:
Extends Word2Vec by representing each word as a bag of character n-grams.
Example: "playing" → "pla", "lay", "ing" + "playing".

Advantages:
Handles Out-of-Vocabulary (OOV) words → can generate embeddings for unseen words by composing from subwords.
Better for morphologically rich languages (German, Finnish, Hindi).

```python
from gensim.models import FastText

sentences = [["I", "love", "playing"], ["players", "are", "winning"]]
model = FastText(sentences, vector_size=50, window=3, min_count=1)
print(model.wv["playing"])
print(model.wv["play"])   # subword helps keep them close
```

Real-World Use Cases
Search engines → query expansion (similar words retrieval).
Recommendation systems → embedding users/items.
Healthcare NLP → embedding clinical terms (ICD/CPT codes).
Fraud detection → embeddings of transaction sequences.

Limitations
Word2Vec & GloVe → produce static embeddings (one vector per word).
"bank" (river vs finance) → same vector.
FastText helps with morphology, but still not contextual.
Solved by → contextual embeddings (ELMo, BERT, GPT).

Interview Tips & Common Questions

Q1. Word2Vec vs GloVe?
Word2Vec: Predictive (neural network, context window).
GloVe: Count-based (matrix factorization, global statistics).

Q2. Word2Vec vs FastText?
Word2Vec: Word-level only.
FastText: Subword-level, handles OOV.

Q3. Why embeddings > TF-IDF?
TF-IDF → sparse, no semantics.
Embeddings → dense, semantic similarity captured.

Q4. Analogy property (very common in interviews):
"king - man + woman ≈ queen".
Be prepared to explain why it works (linear relationships in vector space).

Q5. Why move from static embeddings to contextual embeddings?
Static embeddings can’t handle polysemy (same word, multiple meanings).

# Word Embeddings vs Pretrained Word Embeddings

Word Embeddings (concept)
General technique / idea: representing words as dense vectors in a continuous vector space.
Aim: capture semantic similarity → words with similar meanings should have similar embeddings.
Embeddings can be:
Trained from scratch on your own dataset.
Loaded from pre-trained models trained on large corpora.
So, Word Embeddings = the concept.

Pretrained Word Embeddings
Actual ready-to-use vectors trained on large corpora.
Examples:
Word2Vec (trained by Google on Google News, 100B words)
GloVe (trained by Stanford on Common Crawl, Wikipedia, Twitter)
FastText (trained by Facebook on Wikipedia, Common Crawl, multiple languages)
These are pretrained models of the concept of embeddings.

How they relate
Word2Vec, GloVe, FastText are algorithms/methods to generate word embeddings.
When you download "Google News Word2Vec (300d)" or "GloVe 6B vectors", those are pretrained embeddings using those methods.
You can also run Word2Vec or FastText yourself on your company’s domain-specific corpus → that will produce custom embeddings (not pretrained).

Interview Tip
If asked: “What are Word Embeddings?” →
Say: “They are dense vector representations of words capturing semantic meaning. Algorithms like Word2Vec, GloVe, and FastText can generate them. We often use pretrained embeddings trained on huge corpora, but we can also train embeddings from scratch on domain-specific data.”

So short answer to your question:
Word Embeddings → concept.
Word2Vec, GloVe, FastText → methods/algorithms to create embeddings.
Pretrained Word2Vec/GloVe/FastText models → publicly available embeddings trained on large datasets.

# Sentence & Contextual Embeddings

Sentence Embeddings
Problem
Word embeddings (Word2Vec, GloVe, FastText) give vectors for words, not whole sentences.
"I love NLP" and "NLP is great" → need to be represented close together.
Solution → Sentence Embeddings (map entire sentence → fixed-length dense vector).

Sentence-BERT (SBERT, 2019)
How it works:
Based on BERT/Transformer encoder.
Fine-tuned using Siamese / Triplet networks for semantic similarity tasks.
Produces embeddings where similar sentences are close.

Example:
"How old are you?" ≈ "What is your age?" (close vectors).
"I love NLP" ≠ "I hate NLP" (far vectors).

Real-world use cases:
Semantic search (e.g., query → retrieve similar docs).
Duplicate question detection (Quora, StackOverflow).
FAQ chatbots → match user query to answer.

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')
sentences = ["I love NLP", "NLP is amazing", "I hate maths"]
embeddings = model.encode(sentences)

print(embeddings.shape)  # (3, 384)
```

Universal Sentence Encoder (USE, Google, 2018)

How it works:
Pretrained on large corpora (Wikipedia, QA datasets).

Two architectures:
Transformer version → high accuracy.
DAN (Deep Averaging Network) → faster, lower accuracy.

Output:
512-dim sentence embeddings.

Real-world use cases:
Text clustering
Recommendation engines
Sentiment & intent classification
Multi-language tasks

```python
import tensorflow_hub as hub

model = hub.load("https://tfhub.dev/google/universal-sentence-encoder/4")
sentences = ["I love NLP", "NLP is amazing"]
embeddings = model(sentences)

print(embeddings.shape)  # (2, 512)
```

# Contextual Embeddings

Problem with static embeddings (Word2Vec, GloVe, FastText):

One word = one vector → ignores context.
"bank" in "river bank" vs "money bank" → same embedding.

Solution: Contextual embeddings → embedding depends on surrounding words.

ELMo (Embeddings from Language Models, 2018 – AllenNLP)

How it works:
Based on biLSTM with attention.
Trained as a deep bidirectional language model.
Each word’s embedding is a function of the entire sentence.

Example:
"bank" in "river bank" → embedding close to "water".
"bank" in "money bank" → embedding close to "finance"

Real-world use cases:
Named Entity Recognition (NER).
Sentiment analysis.
Question answering.

```python
# pip install allennlp==0.9 allennlp_models==0.9
from allennlp.commands.elmo import ElmoEmbedder

elmo = ElmoEmbedder()
embeddings = elmo.embed_sentence("I went to the bank to deposit money".split())
print(embeddings.shape)  # (3 layers, 8 words, 1024-dim each)
```

| **Model**          | **Type**        | **Key Idea**                         | **Output**          | **Use Case**                  |
| ------------------ | --------------- | ------------------------------------ | ------------------- | ----------------------------- |
| **Word2Vec/GloVe** | Static word     | One word → one vector                | Word-level (300d)   | Synonym detection, simple NLP |
| **FastText**       | Static word     | Subword embeddings (n-grams)         | Word-level (300d)   | Morphology-rich languages     |
| **ELMo**           | Contextual word | biLSTM + attention, sentence context | Word-level (1024d)  | NER, QA, sentiment            |
| **USE**            | Sentence-level  | Transformer/DAN                      | Sentence (512d)     | Clustering, recommendations   |
| **SBERT**          | Sentence-level  | BERT fine-tuned for semantic sim.    | Sentence (384–768d) | Semantic search, QA, FAQ      |


Interview Tips & Common Questions

Q1. Why SBERT over vanilla BERT embeddings?
Vanilla BERT embeddings are not good for cosine similarity (not optimized).
SBERT fine-tunes BERT specifically for sentence similarity, making it efficient for semantic search.

 Q2. USE vs SBERT?
USE: General-purpose, 512-dim, TensorFlow-based.
SBERT: Optimized for semantic similarity, HuggingFace-based, lighter models exist.

 Q3. Why was ELMo important?
First to bring contextual word embeddings, solving polysemy.
Paved the way for Transformers (BERT, GPT).

 Q4. What’s the difference between Word Embeddings, Sentence Embeddings, Contextual Embeddings?
Word → static, one per word (Word2Vec, GloVe).
Sentence → full sentence vector (SBERT, USE).
Contextual → word vectors depend on context (ELMo, BERT).
