```python
# 2-layer neural network from scratch

x = 1.0
y = 2.0
lr = 0.1

# weights
w1, w2 = 0.5, -0.3
w3, w4 = 0.8, 0.2
b1=b2=b3=0.0

for step in range(5):

    # ---- Forward ----
    h1 = w1*x + b1
    h2 = w2*x + b2

    y_hat = w3*h1 + w4*h2 + b3
    loss = 0.5*(y_hat-y)**2

    print(f"\nStep {step}")
    print("Prediction:", y_hat, "Loss:", loss)

    # ---- Backward ----
    dL_dy = (y_hat-y)

    dL_dw3 = dL_dy*h1
    dL_dw4 = dL_dy*h2

    dL_dh1 = dL_dy*w3
    dL_dh2 = dL_dy*w4

    dL_dw1 = dL_dh1*x
    dL_dw2 = dL_dh2*x

    # ---- Update ----
    w1 -= lr*dL_dw1
    w2 -= lr*dL_dw2
    w3 -= lr*dL_dw3
    w4 -= lr*dL_dw4
```
