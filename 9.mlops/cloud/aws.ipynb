# List S3 buckets

```python
import boto3
from botocore.exceptions import ClientError

def list_s3_buckets():
    # Create an S3 client using your default profile/region
    s3 = boto3.client('s3')
    try:
        response = s3.list_buckets()
    except ClientError as e:
        print(f"Error fetching buckets: {e}")
        return

    buckets = response.get('Buckets', [])
    if not buckets:
        print("No buckets found.")
        return

    print("Your S3 buckets:")
    for b in buckets:
        # b is a dict with 'Name' and 'CreationDate'
        print(f"  - {b['Name']} (created on {b['CreationDate']})")

if __name__ == "__main__":
    list_s3_buckets()
```
