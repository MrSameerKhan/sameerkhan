```python
import cv2
from ultralytics import YOLO

# Load model (choose detection / segmentation / pose)
model = YOLO("/Users/sameerkhan/Desktop/sameerkhan/data/cv/yolov8/yolov8n-pose.pt")

# Input & output paths
input_path = "/Users/sameerkhan/Desktop/sameerkhan/data/cv/yolov8/IMGL8431.jpg"
output_path = "/Users/sameerkhan/Desktop/sameerkhan/data/cv/yolov8/output-IMGL8431.jpg"

# Run inference
results = model(input_path, conf=0.5, verbose=False)

# Annotated frame (numpy array)
annotated = results[0].plot()

# Save output
cv2.imwrite(output_path, annotated)
print(f"Saved annotated image to {output_path}")
```
